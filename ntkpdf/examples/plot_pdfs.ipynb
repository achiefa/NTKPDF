```python
%matplotlib inline
from IPython.display import display

from validphys.pdfgrids import xgrid

from ntkpdf.api import API
from ntkpdf.plotting.pdfplots_providers import DEFAULT_FLAVOURS, DEFAULT_YLABELS
from ntkpdf.plotting.pdfplots_utils import plot_grids
from ntkpdf.utils import XGRID, EVOL_LIST
```

```python
fit = "260526-ac-02-ntk-sgd"
fakepdf = API.fit(fit=fit).as_input()['closuretest']['fakepdf']
```

```python
fakepdf_grid = API.xplotting_grids(pdfs=[fakepdf], Q=1.65, basis="evolution", flavours = EVOL_LIST, xgrid=XGRID)[0]
grids = []
grids.append(API.pdf_grid_at_epoch(fit=fit, epoch=500))
grids.append(API.pdf_grid_at_epoch(fit=fit, epoch=1000))
grids.append(API.pdf_grid_at_epoch(fit=fit, epoch=2000))
grids.append(API.pdf_grid_at_epoch(fit=fit, epoch=5000))
grids.append(fakepdf_grid)
```

```python
gen = plot_grids(
  grids,
  flavours=DEFAULT_FLAVOURS,
  ylabels=DEFAULT_YLABELS,
  labels=["PDF at epoch 500", "PDF at epoch 1000", "PDF at epoch 2000","PDF at epoch 5000", "Fake PDF"],
  # normalise_to=1,
  xscale="linear",
  plot_args=[
      {}, {}, {}, {},
      {"linestyle": "solid", "color":"black", "no_bands": True },
  ]
)
figs = list(gen)

for fig in figs:
    display(fig)
```
